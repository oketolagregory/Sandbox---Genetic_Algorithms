
**Discrete Genetic Algorithm**

Write Python code to randomly generate 10 binary strings with a length of 20 as the population. For each binary string, you are expected to randomly generate 20 values in the range of [0,1] according to uniform distribution, then compare them with 0.5 and get a binary string.

```python
import random

def generate_binary_string_from_uniform_distribution(length):
  """Generates a binary string from a uniform distribution."""
  binary_string = ""
  for i in range(length):
    if random.random() < 0.5:
      binary_string += "0"
    else:
      binary_string += "1"
  return binary_string

def generate_population_from_uniform_distribution(size, length):
  """Generates a population of binary strings from a uniform distribution."""
  population = []
  for i in range(size):
    population.append(generate_binary_string_from_uniform_distribution(length))
  return population

# Generate a population of 10 binary strings with a length of 20.
population = generate_population_from_uniform_distribution(10, 20)

# Print the population.
for binary_string in population:
  print(binary_string)
```

```python
def convert_binary_string_to_decimal(binary_string):
  """Converts a binary string to a decimal number."""
  decimal = 0
  for i in range(len(binary_string)):
    decimal += int(binary_string[i]) * 2**(len(binary_string) - i - 1)
  return decimal

```
