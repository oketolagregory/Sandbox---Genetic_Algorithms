```python
class Particle:
    def __init__(self, route, cost=None):
        # Current route
        self.route = route
        # Personal best route
        self.pbest = route
        # Current cost
        self.current_cost = cost if cost else self.path_cost()
        # Personal best cost
        self.pbest_cost = cost if cost else self.path_cost()
        self.velocity = []

    def clear_velocity(self):
        self.velocity.clear()

    def update_costs_and_pbest(self):
        self.current_cost = self.path_cost()
        if self.current_cost < self.pbest_cost:
            self.pbest = self.route
            self.pbest_cost = self.current_cost

    def path_cost(self):
        return path_cost(self.route)


class PSO:

    # Initialise the Optimisation object with the values supplied
    def __init__(self, iterations, population_size, gbest_probability=1.0, pbest_probability=1.0, cities=None):
        self.cities = cities
        # Global best route
        self.gbest = None
        self.gcost_iter = []
        self.iterations = iterations
        self.population_size = population_size
        self.particles = []
        self.gbest_probability = gbest_probability
        self.pbest_probability = pbest_probability

        solutions = self.initial_population()
        #print("solutions")
        #pprint(solutions)
        # Set up a list of particle objects for each solution (route) in the initial population
        self.particles = [Particle(route=solution) for solution in solutions]

    # Construct a completely random route
    def random_route(self):
        return random.sample(self.cities, len(self.cities))

    def initial_population(self):
        random_population = [self.random_route() for _ in range(self.population_size - 1)]
        greedy_population = [self.greedy_route(0)]
        #  return [*random_population, *greedy_population]
        return [*random_population]
        # return [*greedy_population]

    # Construct a partly optimised route by picking the nearest city at each stage
    def greedy_route(self, start_index):
        unvisited = self.cities[:]
        del unvisited[start_index]
        route = [self.cities[start_index]]
        while len(unvisited):
            index, nearest_city = min(enumerate(unvisited), key=lambda item: item[1].distance(route[-1]))
            route.append(nearest_city)
            del unvisited[index]
        return route

    # This performs the main optimisation
    def run(self):
        # global best is the minimum of the particle costs
        self.gbest = min(self.particles, key=lambda p: p.pbest_cost)
        print(f"initial cost is {self.gbest.pbest_cost}")
        # plt.ion()

        # Draw two figures every 20th iteration
        plt.draw()
        for t in range(self.iterations):
            self.gbest = min(self.particles, key=lambda p: p.pbest_cost)
            if t % 20 == 0:
                # Draw the graph of distance against generation
                fig=plt.figure(0)
                plt.plot(pso.gcost_iter, 'g')
                plt.ylabel('Distance')
                plt.xlabel('Generation')

                fig = plt.figure(0)
                fig.suptitle('pso iter')

                x_list, y_list = [], []
                for city in self.gbest.pbest:
                    x_list.append(city.x)
                    y_list.append(city.y)
                # Last trip returns back to start
                x_list.append(pso.gbest.pbest[0].x)
                y_list.append(pso.gbest.pbest[0].y)

                # Draw the best route found so far
                fig2 = plt.figure(1)
                fig2.clear()
                fig2.suptitle(f'pso TSP iter {t}')
                plt.plot(x_list, y_list, 'ro')
                plt.plot(x_list, y_list, 'g')
                plt.draw()
                plt.pause(0.01)

            # Add best cost to list
            self.gcost_iter.append(self.gbest.pbest_cost)

            for particle in self.particles:
                particle.clear_velocity()
                temp_velocity = []
                gbest = self.gbest.pbest[:]
                new_route = particle.route[:]
                # print("gbest")
                # print(gbest)
                # print("new_route")
                # print(new_route)

                # Set up changes based on the particles personal best position/route
                for i in range(len(self.cities)):
                    if new_route[i] != particle.pbest[i]:
                        swap = (i, particle.pbest.index(new_route[i]), self.pbest_probability)
                        temp_velocity.append(swap)
                        new_route[swap[0]], new_route[swap[1]] = \
                            new_route[swap[1]], new_route[swap[0]]

                # Set up changes based on the swarms global best position/route
                for i in range(len(self.cities)):
                    if new_route[i] != gbest[i]:
                        swap = (i, gbest.index(new_route[i]), self.gbest_probability)
                        temp_velocity.append(swap)
                        gbest[swap[0]], gbest[swap[1]] = gbest[swap[1]], gbest[swap[0]]

                # temp_velocity is a list of swaps to be applied to the new route
                particle.velocity = temp_velocity
                # print("particle.velocity")
                # pprint(particle.velocity)

                # Apply the velocity changes to new route
                # Velocity change means swapping the order of some of the cities
                for swap in temp_velocity:
                    if random.random() <= swap[2]:
                        new_route[swap[0]], new_route[swap[1]] = \
                            new_route[swap[1]], new_route[swap[0]]

                # Update the particle
                particle.route = new_route
                particle.update_costs_and_pbest()


```

```python
import random
import numpy as np

def initialize_particles(num_particles, num_dimensions):
    positions = np.random.randint(0, len(nodes) - 1, size=(num_particles, num_dimensions))

    # Check for repeated nodes and regenerate if necessary
    for particle in positions:
        while not is_valid_path(particle):
            particle = np.random.randint(0, len(nodes) - 1, size=(num_dimensions))

    return positions


def is_valid_path(path):
    path_list = list(path)

    # Check if any node appears multiple times in the path
    for node in path:
        # if path.count(node) > 1:
        if path_list.count(node) > 1:
            return False

    return True


def objective_function(path):
    transmission_rate = 0

    for i in range(len(path) - 1):
        transmission_rate += rates[path[i]].get(path[i + 1], 0)

    # Apply penalty for repeating nodes
    penalty = 0
    for node in path:
        if node in path[:path.index(node) - 1]:
            penalty += 1

    return transmission_rate - penalty


def update_velocity(particle, global_best, personal_best, inertia_weight, c1, c2):
    inertia_term = inertia_weight * particle['velocity']
    cognitive_term = c1 * random.random() * (personal_best['position'] - particle['position'])
    social_term = c2 * random.random() * (global_best['position'] - particle['position'])
    return inertia_term + cognitive_term + social_term


def update_position(particle, velocity):
    new_position = particle['position'] + velocity
    return np.clip(new_position, 0, len(nodes) - 1)


def PSO(num_particles, num_dimensions, max_iterations, c1, c2, inertia_weight):
    particles = [{'position': initialize_particles(1, num_dimensions)[0], 'velocity': 0, 'personal_best': None}
                 for _ in range(num_particles)]
    global_best = {'position': None, 'value': float('-inf')}

    for iteration in range(max_iterations):
        for particle in particles:
            path = [nodes[int(node)] for node in particle['position']]
            value = objective_function(path)

            if particle['personal_best'] is None or value > particle['personal_best']['value']:
                particle['personal_best'] = {'position': particle['position'].copy(), 'value': value}

            if value > global_best['value']:
                global_best = {'position': particle['position'].copy(), 'value': value}

        for particle in particles:
            velocity = update_velocity(particle, global_best, particle['personal_best'], inertia_weight, c1, c2)
            particle['position'] = update_position(particle, velocity)

    optimal_path = [nodes[int(node)] for node in global_best['position']]
    return optimal_path
```
