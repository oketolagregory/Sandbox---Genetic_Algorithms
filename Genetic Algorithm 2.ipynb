```python
# Generate an initial population of routes
def initial_population(population_size):
    print("Initial population")
    # Generate initial random routes from Car-9 to a base station
    initial_routes = []
    # Generate random routes starting from Car-9 to either Base Station-1 or Base Station-2
    for _ in range(population_size):
        route = ['Car9']
        while route[-1] not in base_stations:
            neighbors = list(graph[route[-1]].keys())
            while(True):
              if(len(neighbors) == 0):
                break
              next_node = random.choice(neighbors)
              if(next_node not in route):
                route.append(next_node)
                break
              else:
                neighbors.remove(next_node)
            if(len(neighbors) == 0):
              break
        initial_routes.append(route)
        print("Chromosome " + str(_) + ": " + str(route))
    return initial_routes
```

```python
# Tournament selection to choose parents for crossover
def selection(population, k=5):
    print("Starting tournament selection")
    # Perform selection to choose parents based on fitness
    selected_parents = []
    for _ in range(len(population)):
        participants = random.sample(population, k)
        print("Participants of Group " + str(_) + ": ")
        winner = max(participants, key=lambda route: fitness(route))
        print("Choosen:" + str(winner))
        selected_parents.append(winner)
    return selected_parents
```

```python
# Single-point crossover to create offspring
def crossover(parent1, parent2):
    crossover_point = random.randint(1, min(len(parent1), len(parent2)) - 1)
    child1 = parent1[:crossover_point] + parent2[crossover_point:]
    child2 = parent2[:crossover_point] + parent1[crossover_point:]
    return child1, child2

# Swap mutation to introduce variability
def mutation(route):
    if random.random() < mutation_rate:
        idx1, idx2 = random.sample(range(1, len(route)), 2)
        route[idx1], route[idx2] = route[idx2], route[idx1]
    return route
```
