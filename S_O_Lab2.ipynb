```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range of x values
x = np.linspace(-10, 10, 400)  # Generate 400 evenly spaced points between -10 and 10

# Define the functions
# f1 = x**2

def f1():
  return x **2

def f2():
  return x**3

def f3():
  return x + x**4

def f4():
  return np.exp(x + x**2)

# Plot the functions
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, f1)
plt.title("f(x) = x^2")

plt.subplot(2, 2, 2)
plt.plot(x, f2)
plt.title("f(x) = x^3")

plt.subplot(2, 2, 3)
plt.plot(x, f3)
plt.title("f(x) = x + x^4")

plt.subplot(2, 2, 4)
plt.plot(x, f4)
plt.title("f(x) = e^(x + x^2)")

plt.tight_layout()
plt.show()
```

```python
# Check for minima and maxima
functions = [f1,  f2, f3, f4]
function_names = ["x^2", "x^3", "x + x^4", "e^(x + x^2)" ]

from scipy.optimize import minimize_scalar

for i, func in enumerate(functions):
  result = minimize_scalar(func, bounds=(-10, 10), method="bounded")
  res_type = "minimum" if result.success else "maximum"
  print(f"Function {function_names[i]} has one and only one {res_type} at x = {result.x:.2f}")
```
